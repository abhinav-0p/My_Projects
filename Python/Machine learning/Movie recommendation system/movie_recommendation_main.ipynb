```python
# keep only released, non-adult movies with at least 50 votes
df = df[(df["status"] == "Released") & (df["adult"] == False) & (df["vote_count"] >= 50)]

features = [
    "title",
    "release_date",
    "vote_count",
    "vote_average",
    "poster_path",
    "overview",
    "genres",
    "keywords",
    "cast",
    "directors"
]

movies = df[features].copy()
```

```python
movies.shape
```

```python
movies.isnull().sum()
```

```python
# filling the null value with blanks
text_columns = [
    "overview",
    "genres",
    "keywords",
    "cast",
    "directors"
]

movies[text_columns] = movies[text_columns].fillna("")
```

```python
movies.duplicated().sum()
```

```python
movies = movies.drop_duplicates(subset=["title", "release_date"]).reset_index(drop=True)
```

```python
for col in ["genres", "cast", "directors"]:
    movies[col] = movies[col].apply(
        lambda x: " ".join([name.replace(" ", "").replace("-", "").replace(".", "") for name in x.split(", ")])
    )
```

```python
movies['tag'] = movies['overview'] + ' ' + movies['genres'] + ' ' + movies['keywords'] + ' ' + movies['cast'] + ' ' + movies['directors']
```

```python
movies.drop(columns=['overview','genres','keywords','cast','directors'],inplace=True)
```

```python
import re
import nltk

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.tokenize import TweetTokenizer

nltk.download("stopwords")
```

```python
tags = movies['tag']
```

```python
tags = tags.apply(lambda x: x.lower())
```

```python
import contractions

tags = tags.apply(lambda x: contractions.fix(x))
```

```python
tags = tags.str.replace(r'[^a-zA-Z0-9\s]','',regex=True)
```

```python
tt = TweetTokenizer()
stop = set(stopwords.words('english'))
tags = tags.apply(lambda x: ' '.join([x for x in tt.tokenize(x) if x not in stop]))
```

```python
tags
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfid = TfidfVectorizer(
    stop_words="english",
    min_df=2,
    max_df=0.5
)
vector_tags = tfid.fit_transform(tags)
```

```python
vector_tags.shape
```

```python
import difflib

title_map = dict(zip(movies["title"].str.lower(), movies["title"]))
all_titles = list(title_map.keys())
```

```python
from sklearn.metrics.pairwise import linear_kernel

def recommend(title, n=10):
    matches = movies[movies["title"].str.lower() == title.lower()]

    if matches.empty:
        suggestions = difflib.get_close_matches(title.lower(), all_titles, n=5, cutoff=0.6)
        print("Movie not found.")
        if suggestions:
            print("Did you mean:", [title_map[s] for s in suggestions])
        return

    idx = matches["vote_count"].idxmax()
    scores = linear_kernel(vector_tags[idx], vector_tags).ravel()
    top = scores.argsort()[::-1][1:n+1]
    return movies.iloc[top][["title", "release_date"]]
```

```python
recommend("iron man")
```

```python
import joblib
import scipy.sparse as sp

sp.save_npz("tfidf_matrix.npz", vector_tags)
joblib.dump(movies[["title", "release_date", "poster_path", "vote_average", "vote_count"]], "movies.pkl")
```
